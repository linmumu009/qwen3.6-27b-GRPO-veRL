只读取安全统计。每题型8题；不同题型评分有效分母不同。不把未解决评分计作模型错误。

```python
import json
from pathlib import Path
base = Path('docs') if Path('docs').exists() else Path('.')
data = json.loads((base / 'book_capability_probe_20260909.safe.json').read_text(encoding='utf-8'))
assert data['items'] == data['agreed'] + data['unresolved'] == 40
assert data['prompt_identity_verified'] and not data['training']
assert sum(m['responses'] for m in data['models'].values()) == 160
for row in data['groups']:
    assert row['planned'] == row['agreed'] + row['unresolved']
    assert 0 <= row['closed_full'] <= row['agreed']
    assert 0 <= row['evidence_full'] <= row['agreed']
    if row['split'] == 'all': print(row)
```
