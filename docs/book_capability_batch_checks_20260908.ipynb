# 全书能力型QA安全统计核对
仅核对安全汇总；不读取训练问答、教材或凭据。抽检比例不是总体缺陷率。

```python
import json
from pathlib import Path
base = Path('docs') if Path('docs').exists() else Path('.')
def read(name): return json.loads((base / name).read_text(encoding='utf-8'))
gen = read('book_capability_batch_generation_20260908.safe.json')
rev = read('book_capability_batch_review_20260908.safe.json')
final = read('book_capability_batch_organized_20260908.safe.json')
assert gen['retained_questions'] == rev['before_semantic'] == 453
assert rev['before_semantic'] - rev['semantic_duplicates'] - rev['sample_rejected'] == final['input_candidates']
assert final['input_candidates'] - final['additional_held'] == final['retained_questions'] == 436
assert sum(final['split_counts'].values()) == sum(final['kind_counts'].values()) == 436
assert final['probe_sample_questions'] == 40 and not final['target_probe_completed']
assert not final['training'] and not final['training_ready']
calls = gen['api_calls'] + rev['api_calls'] + rev['previous_failed_semantic_calls']
tokens = gen['usage']['total_tokens'] + rev['usage']['total_tokens'] + rev['previous_failed_semantic_usage']['total_tokens']
assert calls == 337 and tokens == 1020346
print(final['split_counts'], final['kind_counts'], calls, tokens)
```
