# Preflight gate checks
Safe summaries only. Repeatability is not accuracy; final-answer string agreement is not proof of solution validity. Failed gates must not be relabeled as passed.


```python
import json
from pathlib import Path
p=Path('docs') if Path('docs').is_dir() else Path('.')
def read(n): return json.loads((p/f'sft_next_{n}_20260908.safe.json').read_text(encoding='utf-8'))
j=read('single_judge'); d=read('minimal_pairs'); r=read('replay_audit')
assert sum(j['pairs'].values())==j['pair_denominator']==56
assert j['pairs']['joint_agree']==51 and j['gate_passed']
assert j['joint_agreement']==51/56
assert sum(j['score_statuses'].values())==112
assert sum(d['status'].values())==d['attempted']==80
assert d['accepted']==sum(d['accepted_topics'].values())==44
a,b=(d['supervised_tokens_one_exposure'][k] for k in ('a','b'))
assert (a,b)==(3956,6886)
assert d['relative_budget_gap_max_denominator']==abs(a-b)/max(a,b)
assert not d['budget_match_5pct_passed'] and not d['training_ready'] and not d['training_started']
assert r['opensource_rows']==sum(r['datasets'].values())==80
assert r['checks']['source_answer_matches']+r['checks']['answer_needs_review']==80
assert not r['replay_training_ready'] and not r['training_exported']
assert j['usage_tokens']+d['usage_tokens']==1284525
print('PASS: fixed denominators, explicit failed budget/replay gates, no-training state.')
```
