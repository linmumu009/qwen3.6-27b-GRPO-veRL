# Revision40 safe checks
Automatic approval and source-review revisions are different measures, not factual error rates.

```python
import json
from pathlib import Path
p=Path('book_sft_revision40_20260907.safe.json')
if not p.exists(): p=Path('docs')/p
s=json.loads(p.read_text())
assert sum(s['status'].values())==s['completed']==40
assert len({x['id'] for x in s['tasks']})==40
assert sum(s['kind_counts'].values())==40
assert s['review_queue_count']+s['overlap_quarantined']==s['status']['auto_pass']
assert s['usage']['prompt_tokens']+s['usage']['completion_tokens']==s['usage']['total_tokens']
assert not s['ready_for_training']
r=json.loads((p.parent/'book_sft_revision40_source_review8_20260907.safe.json').read_text())
assert sum(r['counts'].values())==len(r['records'])==r['sample_n']==8
assert r['unreviewed_candidates']+r['sample_n']==s['completed']
assert len({x['id'] for x in r['records']})==8
assert all(sum(x['kind']==k for x in r['records'])==2 for k in s['kind_counts'])
assert not r['training_started'] and not r['human_review_complete']
print(r['counts'])
```
