# 六道修订候选预答复核
只读取安全结果；不读取教材、题目、模型回答或凭据。仅统计两轮评分一致项，资料型题不进入闭卷分母。

```python
import json
from pathlib import Path
path = Path('book_repair_value_20260908.safe.json')
if not path.exists(): path = Path('docs') / path
result = json.loads(path.read_text(encoding='utf-8'))
rows = result['results']
assert len(rows) == len({r['id'] for r in rows}) == 6
assert result['generations'] == 20 and result['prompt_identity_verified']
assert not result['training'] and not result['old_pools_modified']
assert sum(r['mode'] == 'closed' for r in rows) == 4
for mode in ('closed', 'evidence'):
    accepted = [r for r in rows if r['mode'] == mode and r['status'] == 'agreed']
    for condition in (('closed', 'evidence') if mode == 'closed' else ('evidence',)):
        for model in ('step120', 'cpt'):
            correct = sum(r['scores'][model + ':' + condition] == 2 for r in accepted)
            print(mode, condition, model, correct, '/', len(accepted))
```
