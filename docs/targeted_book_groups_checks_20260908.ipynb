# 教材概念组安全计数复核
只读取安全汇总，不含教材、问题或答案正文。来源审核通过不等同于训练就绪。

```python
import json
from pathlib import Path
root=Path.cwd()
docs=root if (root/'targeted_book_groups_20260908.safe.json').exists() else root/'docs'
data=json.loads((docs/'targeted_book_groups_20260908.safe.json').read_text(encoding='utf-8'))
assert sum(data['statuses'].values()) == data['groups_generated'] == 120
assert data['groups_accepted'] == sum(data['split_groups'].values()) == sum(data['topic_groups'].values()) == 37
assert data['questions_accepted'] == data['groups_accepted']*3 == 111
assert data['training_ready'] is False
assert data['usage']['prompt_tokens'] + data['usage']['completion_tokens'] == data['usage']['total_tokens']
print({'attempted_groups':120,'accepted_groups':37,'questions':111,'training_ready':False})
probe=json.loads((docs/'targeted_book_probe_20260908.safe.json').read_text(encoding='utf-8'))
selection=json.loads((docs/'targeted_book_selection_20260908.safe.json').read_text(encoding='utf-8'))
pools=json.loads((docs/'targeted_book_pools_20260908.safe.json').read_text(encoding='utf-8'))
assert probe['items']*4 == probe['responses'] == 444
assert probe['finish_reasons'] == {'stop':444}
assert probe['candidate_sha256'] == data['candidates_sha256']
assert sum(selection['status'].values()) == sum(selection['decisions'].values()) == 111
assert pools['pools'] == {'opportunity':63,'maintenance':14,'development':10,'quarantine':24}
assert sum(pools['pools'].values()) == pools['items'] == 111
assert pools['training_ready'] is False and selection['training_started'] is False
print(pools['pools'])
```
